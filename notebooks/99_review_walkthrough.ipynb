# Medical QA LoRA — Reviewed Results Walkthrough

This notebook documents the training evidence already reported by the project. It is a research/portfolio artifact, **not clinical validation**.


## 1. Track the recorded training and validation loss


```python
# Keep train and validation loss side by side so overfitting is easier to spot.
history = [
    (100, 0.6992, 0.6814),
    (200, 0.6724, 0.6647),
    (300, 0.6792, 0.6573),
]
for step, train_loss, val_loss in history:
    print(f"step={step:3d}  train_loss={train_loss:.4f}  val_loss={val_loss:.4f}")
```

### What this tells us

Validation loss improves at every recorded checkpoint, reaching **0.6573** at step 300. Training loss rises slightly from step 200 to 300 while validation loss continues to improve, so the final checkpoint does not show a simple runaway-overfitting pattern from these measurements. Loss alone, however, does not establish medical answer quality.


## 2. Make the parameter-efficiency choice explicit


```python
# LoRA + 4-bit quantization is what makes the experiment practical on a free T4.
print("base_model: Mistral-7B-Instruct-v0.3")
print("training_examples: 5,000")
print("trainable_parameters: 13.6M")
print("reported_trainable_share: 0.36%")
print("quantization: 4-bit QLoRA")
print("hardware: Google Colab T4")
print("reported_cost: $0")
```

### What this tells us

The main engineering contribution is efficient adaptation rather than full-model training. LoRA restricts learning to a small adapter parameter set and 4-bit quantization reduces memory pressure, allowing a multi-billion-parameter model to be adapted on commodity cloud GPU access.
